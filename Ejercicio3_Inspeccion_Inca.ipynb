## Ejercicio 3: Inspección visual de un producto

En una línea de producción, una cámara captura una pequeña imagen de 5×5 píxeles de cada producto, representada como una matriz de NumPy donde cada valor es 0 (píxel normal) o 1 (píxel defectuoso). El agente debe decidir qué hacer con el producto según cuántos píxeles defectuosos detecta.



### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Imagen del producto representada mediante una matriz NumPy de 5×5, donde 0 indica un píxel normal y 1 un píxel defectuoso. |
| **Acciones (A)** | Aprobar el producto, enviarlo a revisión manual o rechazarlo. |
| **Entorno (E)** | Linea de produccion donde se inspecciona el producto con una camara|
| **Objetivo** | Detectar aquellas imagenes que presentan defectos y segun ello tomar una desicion segun los defectos encontrados. |
| **Medida de desempeño (P)** | Clasificar correctamente las imagenes según la cantidad de píxeles defectuosos encontrados. |

### Justificación de las reglas

Para que el sistema o el agente pueda tomar una desicion optima, debemos tener en cuenta las siguientes reglas:
- Si se detecta 0 defectos, entonces la imagen sera aprobada
- Si se detecta entre 1 y 3 defectos, entonces la imagen pasara a revision manual
- Si se dectecta mas de 3 defectos, etnonces la imagen sera rechazada

### Código

Para este punto debemos tener encuenta lo siguiente:
- para que el sistema eliga y tome la mejor desicion la imagen debe estar en una matriz de 5x5
- el valor 1 representa que la imagen tiene defectos y para el valor 0 quiere decir que no se encontro ningun defecto
- La función devuelve una tupla con la accion y motivo

```python
import numpy as np

def agente_inspeccion(imagen):
    defectos = np.sum(imagen == 1) #cuenta cuantos valores son iguales a 1

    if defectos == 0:
        return (
            "Aprobar imagen",
            "No se encontro defectos")
    elif defectos <= 3:
        return (
            "Pasar a revision manual",
            f"Se encontraron {defectos} pixeles defectuosos")
    else:
        return (
            "Rezahar imagen",
            f"Se contraron {defectos} pixeles defectuosos")
```
