## Ejercicio 2: Escalamiento de tickets de soporte técnico

Problema: un sistema de mesa de ayuda decide a qué nivel de soporte enviar un ticket,
considerando:
- tiempo_espera_minutos: un número entero
- nivel_urgencia: un texto, puede ser "baja", "media" o "alta"
- cliente_premium: True o False

Para ello haremos la ficha PEAS



### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Tiempo de espera en minutos, nivel de urgencia y condición de cliente premium. |
| **Acciones (A)** | Enviar el ticket al nivel 1, nivel 2 o nivel 3. |
| **Entorno (E)** | Sistema de mesa de ayuda donde se reciben y atienden tickets |
| **Objetivo** | Asignar cada ticket al nivel de soporte según su urgencia. |
| **Medida de desempeño (P)** | Atender rápidamente los casos urgentes, reducir tiempos de espera excesivos y asignar correctamente los recursos de soporte. |

### Justificación de las reglas
Para este sistema vamos a considerar las siguientes reglas la cual nos serviran para desarrollar el codigo:
- Si la urgencia es alta, el ticket se enviará directamente al nivel 3 sin importar si el cliente es premium o el tiempo de espera
- Si la urgencia es media y el tiempo de espera es mayor a 30 minutos y si el cliente es preium, entonces pasara al nivel 2
- Si la urgencia es media y el tiempo de espera es menor a 30 minutos y el clienete no es preium, entonces pasara al nivel 1
- Si la urgencia es baja y el tiempo de espera es mayor a 60 minutos y el cliente es premium, entonces pasara al nivel 2



### Código

Una ves hecha las reglas vamos a desarrollar el codigo, para ello debemos tener en claro algunas cosas.
- La función recibe tres datos: el tiempo de espera del ticket, el nivel de urgencia y si el cliente es premium.
- Segun la condicion que decida el sistema, el agente debe decidir si el tiket se queda en el nivel 1 o si escala al nivel 2 o al nivel 3 segun la urgencia.
- La funcion debe devolver en una tupla la cual debe tener la accion y el motivo.

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta":
        return (
            "Escalar al nivel 3",
            "Urgencia alta reportada por el cliente")
    elif nivel_urgencia == "media":
        if tiempo_espera_minutos >= 30 or cliente_premium:
            return(
                "Escalar al nivel 2",
                "Urgencia media reportada con tiempo de espera mayor a 30 minutos o cliente premium")
        else:
            return(
                "Escalar al nivel 1",
                "Urgencia reportada con tiempo de espera menor a 30 minutos o no es cliente premium")
    elif nivel_urgencia == "baja":
        if tiempo_espera_minutos >= 60 and cliente_premium:
            return(
                "Escalar al nivel 2 de urgencia",
                "Urgencia baja detectada con cliente preium y tiempo de espera elevada")
        else:
            return(
                "Mantener en el nivel 1",
                "Urgencia baja detectada sin justifacion")
```

### Simulación y pruebas

Para este punto se realizara diferentes pruebas para poder ver los resultados esperados para cada condicion


```python
# Casos de prueba:

pruebas = [
    (5, "alta", False),     # urgencia alta = nivel 3
    (20, "media", False),   # urgencia media, tiempo de espera menor a 30, cliente no premium =  nivel 1
    (30, "media", False),   # urgencia media, tiempo de espera mayor o igual a 30, cliente no premium = nivel 2
    (10, "media", True),    # urgencia media , tiempo de espera menor a 30,  cliente premium = nivel 2
    (20, "baja", False),    # urgencia baja = nivel 1
    (70, "baja", True)      # urgencia baja, tiempo de espera mayor a 30, cliente premium = nivel 2
]

for tiempo, urgencia, premium in pruebas:
    accion, motivo = agente_soporte(tiempo, urgencia, premium)

    print(
        f"Tiempo de espera: {tiempo} min | " f"Urgencia: {urgencia} | " f"Cliente premium: {premium}"
    )

    print(f"Acción: {accion}")
    print(f"Motivo: {motivo}")
    print("-" * 80)
```
